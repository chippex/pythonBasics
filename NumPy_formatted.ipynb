### `np.arange()` and `np.linspace()`

| Function | What it does |
|----------|--------------|
| `np.arange(start, stop, step)` | Numbers from `start` up to (but **not** including) `stop`, jumping by `step` |
| `np.linspace(start, stop, n)` | `n` equally spaced numbers from `start` to `stop` (**both included**) |

````python
import numpy as np

# Stop only
array1 = np.arange(10)
print("array1:", array1)

# Start, Stop, Step
array2 = np.arange(1, 10, 2)
print("array2:", array2)
````

### More `np.arange()` and `np.linspace()` examples

````python
import numpy as np

print(np.arange(5, 20))           # start and stop
print(np.arange(3, 40, 4))        # start, stop, step
print(np.arange(5, 51, 5))        # multiples of 5 up to 50

# linspace: 4 equally spaced numbers from 0 to 200 (both included)
print(np.linspace(0, 200, 4))
````

### `reshape()` — change the shape of an array

`reshape(rows, columns)` rearranges the same elements into a new shape.
The total number of elements must stay the same: `12 = 3 × 4 = 4 × 3 = 2 × 6`.

````python
import numpy as np

grid = np.arange(1, 13)
print(grid)

print(grid.reshape(3, 4))   # 3 rows, 4 columns
print(grid.reshape(2, 6))   # 2 rows, 6 columns
````

## 7️⃣ Sorting

Sorting means arranging data in a specific order (ascending or descending).
`np.sort()` is used to sort the elements of an array.

### Examples

````python
import numpy as np

marks = np.array([79, 80, 56, 99])

# Ascending order
sorted_marks = np.sort(marks)
print(sorted_marks)

# Descending order
sorted_marks = np.sort(marks)[::-1]
print(sorted_marks)
````

````python
names = np.array(["Sayansh Kumar Yadav", "Vaidih", "Arkibh Batman", "North Korea"])

sorted_names = np.sort(names)
print(sorted_names)

sorted_names = np.sort(names)[::-1]
print(sorted_names)
````

### Sorting 2D Arrays

Use the `axis` argument to choose the direction:

| Goal | Argument |
|------|----------|
| Sort **column-wise** (down each column) | `axis = 0` |
| Sort **row-wise** (across each row) | `axis = 1` |

**Syntax**

> `np.sort(array_name, axis = 0)`

````python
a = np.array([[50, 30, 40],
              [60, 20, 80]])

np.sort(a)

np.sort(a, axis=0)
````

````python
arr = np.array([[50, 100, 30],
                [150, 80, 90]])

print(np.sort(arr, axis=0))
print(np.sort(arr, axis=1))
````

### `np.argsort()`

Instead of the sorted values, `argsort()` returns the **positions (indices)** that would sort the array.

````python
a = np.array([80, 70, 11, 100])
print(np.argsort(a))
````

## 8️⃣ Missing Values — `np.isnan()`

`np.isnan(array_name).sum()` counts the missing values (`np.nan`) in an array.

````python
a = np.array([75, np.nan, 80, 90, np.nan])

np.isnan(a).sum()
np.max(a)
````

````python
a = np.array([70, np.nan, 80, 90, np.nan])

print(np.isnan(a))
print(np.isnan(a).sum())
print(np.max(a))
````

### Working with `NaN` values

A normal `mean()` or `max()` returns `nan` if even one value is missing. Use these instead:

| Goal | Code |
|------|------|
| Mask of **missing** values | `np.isnan(a)` |
| Mask of **valid** values | `~np.isnan(a)` |
| Keep only the valid values | `a[~np.isnan(a)]` |
| Average / max / min / sum ignoring `NaN` | `np.nanmean(a)` · `np.nanmax(a)` · `np.nanmin(a)` · `np.nansum(a)` |

````python
import numpy as np

a = np.array([85, np.nan, 74, 92, np.nan, 90])

print("Missing mask:  ", np.isnan(a))
print("Valid values:  ", a[~np.isnan(a)])
print("Plain mean:    ", a.mean())              # nan
print("Mean ignoring NaN:", np.nanmean(a))
print("Max ignoring NaN: ", np.nanmax(a))
print("Min ignoring NaN: ", np.nanmin(a))
````

## 9️⃣ Indexing

**Indexing** means picking out a **single element** from an array using its position.

- Positions start at **0** (the first element is index `0`).
- **Negative** indexes count from the end: `-1` is the last element, `-2` the second last, and so on.

| Array | `78` | `85` | `92` | `66` | `74` |
|-------|:--:|:--:|:--:|:--:|:--:|
| Index | 0 | 1 | 2 | 3 | 4 |
| Negative index | -5 | -4 | -3 | -2 | -1 |

### 1D Array Indexing

````python
import numpy as np

marks = np.array([78, 85, 92, 66, 74])

print("First element: ", marks[0])
print("Third element: ", marks[2])
print("Last element:  ", marks[-1])
print("Second last:   ", marks[-2])
````

### 2D Array Indexing

In a 2D array use **two** indexes, `array[row, column]`.

> `scores[row, column]`

````python
# 5 students x 2 subjects (Math, Science)
scores = np.array([
    [78, 82],
    [85, 79],
    [92, 95],
    [66, 70],
    [74, 68]
])

print("Row 0, Column 1:  ", scores[0, 1])    # first student, Science
print("Row 2, Column 0:  ", scores[2, 0])    # third student, Math
print("Last row, last col:", scores[-1, -1])
````

### Changing a value using its index

An index can also be used to **update** an element.

````python
marks = np.array([78, 85, 92, 66, 74])

marks[3] = 70      # change the 4th mark from 66 to 70
print(marks)
````

## 🔟 Slicing

**Slicing** means picking out a **range** of elements (a part of the array).

> `array[start : stop : step]`

| Part | Meaning | Default |
|------|---------|---------|
| `start` | Index where the slice begins (**included**) | `0` |
| `stop` | Index where the slice ends (**not included**) | end of array |
| `step` | Jump between elements | `1` |

### 1D Array Slicing

````python
import numpy as np

nums = np.array([10, 20, 30, 40, 50, 60, 70])

print("nums[1:4]  ->", nums[1:4])     # index 1, 2, 3
print("nums[:3]   ->", nums[:3])      # from the start up to index 2
print("nums[4:]   ->", nums[4:])      # from index 4 to the end
print("nums[::2]  ->", nums[::2])     # every 2nd element
print("nums[::-1] ->", nums[::-1])    # reversed
````

### 2D Array Slicing

Give a slice for the rows and a slice for the columns, separated by a comma.

> `array[row_slice, column_slice]`

A single `:` on its own means **"all"**.

````python
scores = np.array([
    [78, 82],
    [85, 79],
    [92, 95],
    [66, 70],
    [74, 68]
])

print("First 2 students:")
print(scores[0:2])

print("\nAll Math marks (column 0):")
print(scores[:, 0])

print("\nAll marks of student 2 (row 1):")
print(scores[1, :])

print("\nStudents 1 to 3, Science only:")
print(scores[1:4, 1])
````

### Quick reference

| Goal | Code |
|------|------|
| One row | `arr[1, :]` |
| One column | `arr[:, 1]` |
| First two rows | `arr[:2]` |
| Every alternate row | `arr[::2]` |
| Reverse all rows | `arr[::-1]` |
| Reverse the columns | `arr[:, ::-1]` |
| A block (rows 1–2, columns 2–3) | `arr[1:3, 2:4]` |

> ⚠️ **Slices are views, not copies.** Changing a slice also changes the original array. Use `.copy()` if you want an independent copy.

````python
a = np.array([1, 2, 3, 4, 5])

view = a[1:4]
view[0] = 99
print("Original after changing the slice:", a)    # a changed too!

b = np.array([1, 2, 3, 4, 5])
safe = b[1:4].copy()
safe[0] = 99
print("Original after changing the copy: ", b)    # b is unchanged
````

## 1️⃣1️⃣ Filtering (Boolean Indexing)

**Filtering** means selecting only the elements that satisfy a **condition**.

1. A condition such as `marks > 80` is checked for every element and gives an array of `True` / `False` (a *mask*).
2. Putting the mask inside the square brackets keeps only the `True` elements.

> `array[condition]`

| Operator | Meaning |
|:--:|---------|
| `>` `<` `>=` `<=` | greater / less than |
| `==` `!=` | equal / not equal |
| `&` | **and** — both conditions must be true |
| `\|` | **or** — at least one condition is true |
| `~` | **not** — reverses the mask |

### Step 1 — the Boolean mask

````python
import numpy as np

marks = np.array([35, 67, 42, 89, 28, 76, 51, 39])

mask = marks > 50
print(mask)
````

### Step 2 — use the mask to filter

````python
print("Marks above 50:", marks[marks > 50])
print("Marks 40 or above:", marks[marks >= 40])
````

### Counting matches

`True` counts as `1`, so `.sum()` on a mask counts how many elements pass the condition.

````python
print("How many scored above 50?", (marks > 50).sum())
print("Same answer using len():  ", len(marks[marks > 50]))
````

### Combining conditions

Put **each condition in brackets** when using `&` or `|`.

````python
# Marks between 40 and 80
print("AND:", marks[(marks >= 40) & (marks <= 80)])

# Very low or very high marks
print("OR: ", marks[(marks < 30) | (marks > 85)])

# NOT: everything that is NOT above 50
print("NOT:", marks[~(marks > 50)])
````

### `np.where()` — choose a value based on a condition

> `np.where(condition, value_if_true, value_if_false)`

````python
result = np.where(marks >= 40, "Pass", "Fail")
print(marks)
print(result)
````

### Filtering 2D arrays

````python
scores = np.array([
    [78, 82],
    [85, 79],
    [92, 95],
    [66, 70],
    [74, 68]
])

# Every individual mark above 80 (result is a flat 1D array)
print("Marks above 80:", scores[scores > 80])

# Whole rows: students who scored above 80 in Math (column 0)
print("\nStudents with Math above 80:")
print(scores[scores[:, 0] > 80])
````

### Even and odd numbers with `%`

`%` gives the **remainder** after division. A number is even if the remainder when divided by 2 is `0`, and odd otherwise.

````python
import numpy as np

ages = np.array([[12, 34, 23, 45, 67, 66, 13, 15],
                 [18, 23, 99, 22, 52, 14, 17, 90]])

print("Even:", ages[ages % 2 == 0])
print("Odd: ", ages[ages % 2 != 0])
````

### Grouping data with conditions

````python
teenagers = ages[ages < 18]
adults    = ages[(ages >= 18) & (ages < 65)]
citizens  = ages[ages >= 65]

print("Teenagers:", teenagers)
print("Adults:   ", adults)
print("Senior citizens:", citizens)
````

### `np.where()` keeps the original shape

Filtering with `array[condition]` gives a flat list of matches. `np.where()` keeps the **same shape** and replaces the elements that fail the condition.

````python
# Keep ages 18 or above, replace the rest with 0
print(np.where(ages >= 18, ages, 0))
````

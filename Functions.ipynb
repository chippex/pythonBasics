# 📘 Introduction to Functions

A **function** is a reusable block of code that performs a specific task.

Instead of writing the same code multiple times, you can write it once inside a function and call it whenever needed.

**Key Characteristics**
- Reusable
- Improves readability
- Reduces repetition
- Makes code easier to maintain

## Defining a Function

A function is created using the `def` keyword.

**Syntax**

```python
def function_name():
    # Code
```

````python
def greet():
    print("Hello!")

greet()
````

## Calling a Function

A function executes **only when it is called**.

To call a function, write its name followed by parentheses `()`.

````python
def greet():
    print("Welcome to Python!")

greet()
greet()
````

## Function with Parameters

Parameters allow values to be passed into a function.

**Syntax**

```python
def function_name(parameter):
    # Code
```

````python
def greet(name):
    print("Hello", name)

greet("Alex")
greet("John")
````

## Function with Multiple Parameters

A function can accept more than one parameter.

````python
def add(a, b):
    print(a + b)

add(5, 10)
````

## Returning a Value

Use the `return` keyword to send a value back from a function.

````python
def square(num):
    return num * num

result = square(5)

print(result)
````

## Difference Between `print()` and `return`

| `print()`                 | `return`                           |
| ------------------------- | ---------------------------------- |
| Displays the output       | Sends the value back to the caller |
| Cannot be stored directly | Can be stored in a variable        |
| Ends after displaying     | Allows further operations          |


````python
def square(num):
    return num * num

result = square(6)

print(result + 4)
````

## Benefits of Functions

* Reuse code
* Reduce repetition
* Improve readability
* Make debugging easier
* Organize programs into smaller parts

````python
#Write a Python function to find the larger of two numbers entered by the user.

num1 = int(input("Enter first number for comparison: ")) #Taking values from the user.
num2 = int(input("Enter second number for comparison: "))

def comparison(num1,num2): #Defining the function and parameters.

  if num1 > num2:
    print(f"{num1} is greater than {num2}.") #Comparison of both numbers with if, elif statements.

  elif num1 < num2:
    print(f"{num1} is lesser than {num2}.")

  elif num1 == num2:
    print("Both numbers are equal.")

comparison(num1,num2) #Calling the function with arguments in brackets.
````

````python
# Write a function to check whether a number is positive, negative, or zero.

num = int(input("Enter number: ")) # Taking integer input from the user.

def positiveOrNegative(num): # Defining the function to check the number's sign.

  if num > 0:
    print(f"{num} is a positive integer.") # If the number is greater than 0, it's positive.

  elif num < 0:
    print(f"{num} is a negative integer.") # If the number is less than 0, it's negative.

  else:
    print(f"{num} is zero.") # If the number is neither positive nor negative, it must be zero.

positiveOrNegative(num) # Calling the function with the user-provided number.
````

````python
"""
Write a Python program that uses a function named total_plants to calculate the total number of saplings and flowerpots distributed during the Green Earth Campaign.
The function should add the values and print the result.
"""

saplingsDistributed = int(input("Enter number of saplings distributed during Green Earth Campaign: ")) # Taking input for the number of saplings.
flowerpotsDistributed = int(input("Enter number of flowerpots distributed during Green Earth Campaign: ")) # Taking input for the number of flowerpots.

def total_plants(saplingsDistributed, flowerpotsDistributed): # Defining the function to calculate the total.

  total_plants = saplingsDistributed + flowerpotsDistributed # Summing the saplings and flowerpots.
  print(f"{total_plants} saplings and flowerpots were distributed during the Green Earth Campaign.") # Printing the total distributed items.

total_plants(saplingsDistributed, flowerpotsDistributed) # Calling the function with the provided inputs.
````

````python
'''
Avi has some pencils. He buys some more pencils. Write a Python function to determine whether the total number of pencils is even or odd.
First, take the number of pencils Avi has and the number of pencils he buys from the user, then calculate the total.
'''

aviPencils = 6 # Initial number of pencils Avi has.
aviBuys = int(input("How many pencils does Avi buy? ")) # Number of pencils Avi buys, taken as input from the user.

def even_or_odd(initial_pencils, bought_pencils): # Defining a function to check if the total is even or odd.

  totalPencils = initial_pencils + bought_pencils # Calculating the total number of pencils.

  if totalPencils % 2 == 0:
    print(f"The total number of pencils ({totalPencils}) is even.") # If the remainder when divided by 2 is 0, it's even.

  else:
    print(f"The total number of pencils ({totalPencils}) is odd.") # Otherwise, it's odd.

even_or_odd(aviPencils, aviBuys) # Calling the function with Avi's initial pencils and bought pencils.
````

````python
'''
Ayra has 15 books at home and receives 5 more books as a gift. Write a Python function to check whether the total number of books is a multiple of 5.
What is the total number of books Ayra‣having?‣
Is the total number of books a multiple of 5?
'''

ayraBooksHome = 15 # Number of books Ayra has at home.
ayraBooksGifted = 5 # Number of books Ayra receives as a gift.

def totalBooks(ayraBooksHome,ayraBooksGifted): # Defining a function to calculate total books and check if it's a multiple of 5.

  totalBooks = ayraBooksHome + ayraBooksGifted # Calculating the total number of books.
  print(f"Total amount of books that Ayra has is {totalBooks}") # Printing the total number of books.

  if totalBooks % 5 == 0: # Checking if the total number of books is a multiple of 5.
    print("Total amount of books is a multiple of 5.") # If it is, print this message.

  else:
    print("Total amount of books is not a multiple of 5.") # If it's not, print this message.

totalBooks(ayraBooksHome,ayraBooksGifted) # Calling the function with Ayra's home books and gifted books.
````

# 📘 Introduction to Tuples

A **tuple** is an **ordered** and **immutable** collection of elements in Python.

**Key Characteristics**
- Ordered
- Immutable
- Allows duplicate values
- Indexed
- Written using `()`

## Creating Tuples

```python
# Multiple elements
fruits = ("Apple", "Banana", "Mango")
print(fruits)
```

## Single Element Tuple

A single element tuple **must** contain a trailing comma `,`.

```python
fruit = ("Apple",)
print(type(fruit))
```

## Empty Tuple

```python
numbers = tuple([1, 2, 3])
letters = tuple("Python")

print(numbers)
print(letters)
```

## Creating Tuples using `tuple()`

```python
numbers = tuple([1, 2, 3])
letters = tuple("Python")

print(numbers)
print(letters)
```

## Adding Elements

Tuples are **immutable**, so they cannot be modified directly.

Convert the tuple into a list, modify it, then convert it back into a tuple.

```python
numbers = (1, 2, 3)

temp = list(numbers)
temp.append(4)

numbers = tuple(temp)

print(numbers)
```

## Concatenation (`+`)

Combines two tuples into one.

```python
a = (1, 2)
b = (3, 4)

print(a + b)
```

## Repetition (`*`)

Repeats the tuple multiple times.

```python
colors = ("Red",)

print(colors * 3)
```

## Indexing

```python
colors = ("Red", "Green", "Blue")

print(colors[0])
print(colors[-1])
```

## Slicing

```python
colors = ("Red", "Green", "Blue", "Yellow", "Pink")

print(colors[1:4])
```

## Reverse Slicing

```python
print(colors[::-1])
```

## Advanced Slicing

```python
print(colors[::2])
```

## Membership Operator (`in`)

```python
colors = ("Red", "Green", "Blue")

print("Green" in colors)
```

## Looping Through a Tuple

```python
for color in colors:
    print(color)
```

## `max()`

```python
numbers = (10, 30, 5, 70)

print(max(numbers))
```

## `count()`

Returns the number of occurrences of an element.

```python
numbers = (1, 2, 2, 2, 3)

print(numbers.count(2))
```

## `index()`

Returns the index of the first occurrence of an element.

```python
numbers = (10, 20, 30)

print(numbers.index(20))
```

## Packing

```python
student = ("Alex", 14, "A")
```

## Unpacking

```python
name, age, grade = student

print(name)
print(age)
print(grade)
```

## Unpacking with `*`

```python
a, *b, c = (1, 2, 3, 4, 5)

print(a)
print(b)
print(c)
```

## Deleting a Tuple

```python
numbers = (1, 2, 3)

del numbers
```

# 📋 Summary

| Topic            | Syntax               | Purpose                               |
| ---------------- | -------------------- | ------------------------------------- |
| Create Tuple     | `t = (1, 2, 3)`      | Creates a tuple                       |
| Empty Tuple      | `()`                 | Creates an empty tuple                |
| Single Element   | `(5,)`               | Creates a single-element tuple        |
| Convert to Tuple | `tuple(list)`        | Converts an iterable into a tuple     |
| Length           | `len(t)`             | Returns number of elements            |
| Concatenation    | `a + b`              | Joins two tuples                      |
| Repetition       | `t * 3`              | Repeats a tuple                       |
| Indexing         | `t[0]`               | Access an element                     |
| Slicing          | `t[1:4]`             | Returns a portion of the tuple        |
| Reverse          | `t[::-1]`            | Reverses the tuple                    |
| Step Slicing     | `t[::2]`             | Returns every nth element             |
| Membership       | `x in t`             | Checks if an element exists           |
| Loop             | `for x in t:`        | Iterates through a tuple              |
| Maximum          | `max(t)`             | Returns the largest value             |
| Count            | `t.count(x)`         | Counts occurrences                    |
| Index            | `t.index(x)`         | Returns the first index of an element |
| Packing          | `t = (1, "A", True)` | Packs values into a tuple             |
| Unpacking        | `a, b, c = t`        | Assigns tuple values to variables     |
| Star Unpacking   | `a, *b, c = t`       | Collects remaining values into a list |
| Delete           | `del t`              | Deletes the entire tuple              |

## **1. Tuple Treasure Hunt**

```python
clues = ("Start", "Go left", "Find key", "Open door", "Treasure!")
steps = (5, 3, 8, 2, 1)
extra_clues = ("Watch out", "Hidden trap")
repeat_clues = ("Look", "Search")

#a
print("There are",(len(clues + extra_clues + repeat_clues)),"clues in total.")

#b
if "Find key" in clues:
  print("'Find key' is in clues.")

#c
print("The highest number in steps is",(max(steps)))

#d
print("The lowest number in steps is",(min(steps)))

#e
print(clues + extra_clues)

#f
print(clues * 2)
```

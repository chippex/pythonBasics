###**Dictionaries**

A Python dictionary is a built-in data structure used to store data in `key:value` pairs. It is mutable (changeable), does not allow duplicate keys, and preserves the insertion order of items.

---

`{}` - curly braces **enclose** the elements

`:` - colon **seperates** keys from values

`,` - comma **seperates** `key:value` pairs

---

##**Characteristics**

1. **Ordered**
> Maintains insertion order

2. **Mutable**
> Items can be added, changed, or removed after creation

3. **No Duplicate Keys**
> Each key must be unique; duplicate keys silently overwrite the earlier value


##Creating Dictionaries

Empty dictionaries using `{}` and `dict()`

````python
#using curly braces
dict1 = {}

print(dict1)

print(type(dict1))
````

````python
#using dict() constructor
dict1 = dict()

print(dict1)

print(type(dict1))
````

Creating non - empty dictionary

Syntax:

```
dict = {key1: value1, key2: value2, key3: value3}
```



````python
# Using curly braces (Most Common)
user = {
    "name": "Alice",
    "age": 25,
    "is_admin": True
}

# Using the dict() constructor
user_2 = dict(name="Bob", age=30, is_admin=False)

print(user)
print(user_2)
````

###Practice
Creating dictionaries with `{}` and `dict()`


---

Scenario
Your school library wants to store book titles and the number of copies available.



1.   Create with `{}`

    create a dictionary called library using curly braces with:
    "Harry Potter" -> 5
    "Matilda" -> 3
    "The Hobbit" -> 5
2.   Create with `dict()`

    Also create the same dictionary using the dict() constructor.
    Note: with dict(), keys with spaces must use the {} approach or be passed differently.

3.  Print both dictionaries, then check if they are equal using `==` and print the result



````python
library = {
    "Harry Potter": 5,
    "Matilda": 3,
    "The Hobbit": 5
}

library_dict_constructor = dict([
    ("Harry Potter", 5),
    ("Matilda", 3),
    ("The Hobbit", 5)
])

print("Dictionary created with curly braces:", library)
print("Dictionary created with dict() constructor:", library_dict_constructor)
print("Are the two dictionaries equal?", library == library_dict_constructor)
````
